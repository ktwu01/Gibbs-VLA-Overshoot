# Gibbs–VLA visualization

Use this notebook to plot trajectories, FFT/Welch spectra, wavelet scalograms, and Gibbs overshoot metrics. Treat robot data as multivariate time series (`(T, D)`).

```python
import numpy as np
import matplotlib.pyplot as plt

from analysis.metrics import calculate_gibbs_overshoot, spectral_analysis
from analysis.fourier import fft_magnitude, power_spectral_density

# Example: synthetic step + ringing
t = np.linspace(0, 1, 500)
step = np.where(t > 0.5, 1.0, 0.0)
ringing = 0.09 * np.sin(2 * np.pi * 40 * t) * np.exp(-15 * np.abs(t - 0.5))
x = step + ringing

fig, ax = plt.subplots()
ax.plot(t, x)
ax.set_title("Synthetic step + transient")
plt.show()

freqs, psd = power_spectral_density(x, fs=500)
plt.figure()
plt.semilogy(freqs, psd)
plt.xlabel("Hz")
plt.ylabel("PSD")
plt.show()
```
